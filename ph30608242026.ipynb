```python
import numpy as np
a = np.arange(1, 1001)
b = np.linspace(-7, 8, 1000)

def add_array(a, b):
  c = np.empty_like(a)
  for i in range(len(a)):
    c[i] = a[i] + b[i]
  return np.array(c)
```

```python
%%timeit
add_array(a, b)
# this variation is marginally quicker but still incredibly slow
```

```python
from matplotlib import pyplot as plt
```

```python
# Vars
x = np.linspace(-5, 5, 101) # keep odd array sizes, as without u dont count the ends (0 - 100)
y = -5 + 2*x + x**2


# cardinal sin is using the plt command over and over again, limiting what you can actually do
# the plt command center is just for quick and dirty checks
plt.plot(x, y)
plt.xlabel('x (s)')
plt.ylabel('y (m)')
plt.show()
```

```python
# Get a fig/ax
# now we have control over the figure, where the axes and the drawing actually exist
# fig/ax objects let u control the entire figure
fig, ax = plt.subplots()
ax.plot(x, y)
ax.set_xlabel('x (s)')
ax.set_ylabel('y (m)')
plt.show()

# more typing but u get more control over each object within the axes
```
