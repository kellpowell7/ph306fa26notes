4. approximation error

6. roundoff error

ex: 1/3 = 0.3 repeating, eventually the computer has to round off somewhere, so there is some error induced when the computer has to round off.

```python
from functools import lru_cache
import numpy as np
from numba import jit, njit
```

```python
@lru_cache
def factorial(n):
  if n == 0:
      return 1
  else:
    return n * factorial(n-1)
# basic recursive function
# lru cache shows that we implicitly had to solve 99!, 98! etc, and it stores the answer in advance
```

```python
%%timeit
factorial(100)
```
